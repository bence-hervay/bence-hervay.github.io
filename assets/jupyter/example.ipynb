```python
# Even Fibonacci numbers below four million (Project Euler #2)
a, b, total = 1, 2, 0
while b < 4_000_000:
    if b % 2 == 0:
        total += b
    a, b = b, a + b
total
```

```python
# The ratio the sequence converges to, for good measure.
print('ratio ->', b / a)
```
